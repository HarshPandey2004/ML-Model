```python
import numpy as np
import pandas as pd
```

```python
employee = pd.DataFrame({
    'emp_id' : [1, 2, 3, 4,5],
    'emp_name' : ['John', 'Smith', 'Alice', 'Bob','David'],
    'dept_id' : [101, 102, 101, 103,104]
})
#dataframe 2 salary information
salary = pd.DataFrame({
    'emp_id' : [1, 2, 3, 6,7],
        'salary' : [5000, 6000, 7000, 8000,9000],
        'bonus' : [500, 600, 700, 800,900]
})
```

```python
employee
```

```python
salary
```

```python
pd.merge(employee, salary)
```

```python
pd.merge(employee, salary, on='emp_id', how='inner')  # Inner join
#we use inner join to get only the records that have matching values in both dataframes.
```

```python
pd.merge(employee, salary, on='emp_id', how='outer')  # Outer join
#we use outer join to get all the records from both dataframes. If there is no match, it will fill with NaN.
```

```python
df1 = pd.DataFrame({
    'A' : ['A3', 'A4', 'A5', 'A6'],
    'B' : ['B3', 'B4', 'B5', 'B6'],
    'C' : ['C3', 'C4', 'C5', 'C6'],
    'D' : ['D3', 'D4', 'D5', 'D6']
})

df2 = pd.DataFrame({
    'A' : ['A0', 'A1', 'A2', 'A3'],
    'B' : ['B0', 'B1', 'B2', 'B3'],
    'C' : ['C0', 'C1', 'C2', 'C3'],
    'D' : ['D0', 'D1', 'D2', 'D3']
})
```

```python
a1 = pd.DataFrame({
    'A' : ['A0', 'A1', 'A2', 'A3'],
    'B' : ['B0', 'B1', 'B2', 'B3'],
    'C' : ['C0', 'C1', 'C2', 'C3'],
    'D' : ['D0', 'D1', 'D2', 'D3']
})

a2 = pd.DataFrame({
    'A' : ['A4', 'A5', 'A6', 'A7'],
    'B' : ['B4', 'B5', 'B6', 'B7'],
    'C' : ['C4', 'C5', 'C6', 'C7'],
    'D' : ['D4', 'D5', 'D6', 'D7']
})
```

```python
a1.join(a2, lsuffix='_left', rsuffix='_right')  # Joining along columns
```
