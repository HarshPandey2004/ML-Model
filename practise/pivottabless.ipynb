```python
import numpy as np
import pandas as pd
```

```python
data = {
    'Date': pd.date_range('2023-01-01', periods=6,),
    'Product': ['A', 'B', 'A', 'B', 'A', 'B'],
    'Region': ['North', 'North', 'South', 'South', 'East', 'East'],
    'Sales': np.random.randint(100, 500, size=6),
    'Units': np.random.randint(1, 10, size=6),
    'Rep': ['John', 'Jane', 'John', 'Jane', 'John', 'Jane']
}

df = pd.DataFrame(data)
df['Month'] = df['Date'].dt.month
df['Year'] = df['Date'].dt.year
```

```python
df = pd.DataFrame(data)
df
```

```python
pd.pivot_table(df, values='Sales', index=['Region'],columns=['Product'], aggfunc=np.sum )
```
