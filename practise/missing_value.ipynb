```python
import pandas as pd
import numpy as np
```

```python
#Find missing values in a dataset
data ={
    'A' : [1, 2, np.nan, 4],
    'B' : [np.nan, 2, 3, 4],
    'C' : [1, 2, 1, 4],
    'D' : [np.nan, 2, np.nan, np.nan]
}
dff =pd.DataFrame(data)
dff.isna()  # Checking for missing values in the DataFrame
dff.isna().sum()  # Counting the number of missing values in each column
```

```python
dff.dropna(thresh=1)  # Dropping rows with missing values
```

```python
dff.fillna(0,inplace=True)  # Filling missing values with 0
#we write inplace=True to make changes in the original DataFrame. 
#If we don't write inplace=True, it will return a new DataFrame with missing values filled with 0,
#but the original DataFrame will remain unchanged.
```

```python
values = {'A' : 0, 'B' : 10, 'C' : 100, 'D' : 1000}
dff.fillna(value=values, inplace=True)  # Filling missing values with different values for each column
```
