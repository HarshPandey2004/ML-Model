```python
data ={
    'Name' : ['Alice', 'Bob', 'Charlie'],
    'Age' : [25, 30, 35],
    'City'  : ['New York', 'Los Angeles', 'Chicago'],
    'Salary' : [50000, 60000, 70000]
}

df = pd.DataFrame(data)
df
```

```python
data_list = [
    ['Alice', 25, 'New York', 50000],
    ['Bob', 30, 'Los Angeles', 60000],
    ['Charlie', 35, 'Chicago', 70000]
]
df2 = pd.DataFrame(data_list)
df2

```
