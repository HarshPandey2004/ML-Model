```python
import pandas as pd
import numpy as np
```

```python
data = {
    'Company': ['GOOGLE', 'GOOGLE', 'MSFT', 'MSFT', 'FB', 'FB'],
    'Person': ['Sam', 'Charlie', 'Amy', 'Vanessa', 'Carl', 'Sarah'],
    'Sales': [200, 120, 340, 124, 243, 350],
    'Date': pd.date_range('2023-01-01', periods=6,)   
}
```

```python
df =pd.DataFrame(data)
df
```

```python
df.groupby('Company')['Sales'].sum()
#this will group the data by the 'Company' column and then sum the 'Sales' for each company.
```

```python
df['Sales'].aggregate(['sum', 'mean', 'std'])
```
